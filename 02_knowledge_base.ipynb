```python
import boto3
from dotenv import load_dotenv
import os
import json

load_dotenv(override=True)

aws_access_key_id = os.getenv("AWS_ACCESS_KEY_ID")
aws_secret_access_key = os.getenv("AWS_SECRET_ACCESS_KEY")
aws_region = os.getenv("AWS_REGION")
aws_account_id = os.getenv("AWS_ACCOUNT_ID")

sesion_aws = boto3.Session(
    aws_access_key_id=aws_access_key_id,
    aws_secret_access_key=aws_secret_access_key,
    region_name=aws_region
)

client = sesion_aws.client('bedrock-agent-runtime')
```

```python
import boto3
import json
from botocore.exceptions import ClientError

# Parámetros que debes ajustar
aws_region = "us-east-1"
kb_id = "KNOWLEDGE_BASE_ID_AQUI"


response = client.retrieve(
    knowledgeBaseId=kb_id,
    retrievalQuery={
        "text": "¿Cuáles son las partes del motor?"
    },
    retrievalConfiguration={
        "vectorSearchConfiguration": {
            "numberOfResults": 5,
            "overrideSearchType": "SEMANTIC",
        }
    }
)

print("=== 🔎 Resultados de retrieve ===")
for doc in response["retrievalResults"]:
    print("\nTexto:")
    print(doc["content"]["text"])
    print("Fuente:", doc["location"]["s3Location"]["uri"])
    print("Score:", doc["score"])
```

```python
response = client.retrieve_and_generate(
    input={
        "text": "dame la licencia de Marquina Garcia"
    },
    retrieveAndGenerateConfiguration={
        "type": "KNOWLEDGE_BASE",
        "knowledgeBaseConfiguration": {
            "knowledgeBaseId": kb_id,
            "modelArn": "arn:aws:bedrock:us-east-1::foundation-model/amazon.nova-pro-v1:0",
            "retrievalConfiguration": {
                "vectorSearchConfiguration": {
                    "numberOfResults": 3,
                    "overrideSearchType": "SEMANTIC",
                }
            },
            "generationConfiguration": {
                "inferenceConfig": {
                    "textInferenceConfig": {
                        "temperature": 0.7,
                        "maxTokens": 300
                    }
                },
                "performanceConfig": {
                    "latency": "standard"
                }
            }
        }
    }
)

print("=== 🧠 Respuesta generada ===")
print(response["output"]["text"])
print("\n=== 🔍 Citas (sources) ===")
print(json.dumps(response.get("citations", []), indent=2))
```
